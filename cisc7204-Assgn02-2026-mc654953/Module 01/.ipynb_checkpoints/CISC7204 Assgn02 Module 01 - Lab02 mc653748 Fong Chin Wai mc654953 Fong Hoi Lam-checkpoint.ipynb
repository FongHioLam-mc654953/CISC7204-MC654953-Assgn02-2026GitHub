# Introduction to Matplotlib and Line Plots

In this lab, we learn how to create **line plots** with Matplotlib, using the cleaned Canada immigration dataset (`Canada.csv`).

```python
import pandas as pd
import numpy as np

print('pandas version:', pd.__version__)
```

## Read the Cleaned Dataset

Lab 02 uses the **pre-processed** dataset `Canada.csv` (the output of Lab 01 pre-processing).

```python
df_can = pd.read_csv('Canada.csv')

df_can.head()
```

```python
df_can.set_index('Country', inplace=True)

df_can.head(3)
```

## Create a List of Years

Create a list of the years from 1980 to 2013 as strings.

```python
years = list(map(str, range(1980, 2014)))

years[:5]
```

## Import Matplotlib

```python
%matplotlib inline

import matplotlib as mpl
import matplotlib.pyplot as plt

print('Matplotlib version:', mpl.__version__)
```

```python
# Optional: apply the ggplot style for nicer plots
plt.style.use('ggplot')

print('Style applied: ggplot')
```

### Case 1: Immigration from Haiti (2010 Earthquake)

Haiti suffered a devastating earthquake in 2010. Let's visualize the immigration trend and annotate the earthquake year.

```python
haiti = df_can.loc['Haiti', years]

haiti.plot()
plt.title('Immigration from Haiti')
plt.ylabel('Number of Immigrants')
plt.xlabel('Year')
plt.show()
```

```python
# Convert the index to integers for annotation
haiti.index = haiti.index.map(int)

haiti.plot(kind='line')
plt.title('Immigration from Haiti')
plt.ylabel('Number of Immigrants')
plt.xlabel('Year')
plt.text(2010, haiti[2010], '2010 Earthquake')  # annotate the earthquake
plt.show()
```

### Exercise 1: Compare the immigration trends of **India** and **China** from 1980 to 2013.

```python
df_can.loc[['India', 'China'], years].transpose().plot(kind='line', figsize=(10, 6))
plt.title('Immigration from India and China (1980-2013)')
plt.ylabel('Number of Immigrants')
plt.xlabel('Year')
plt.show()
```

### Exercise 2: Plot the trends of the **top 5 countries** that immigrated the most to Canada.

```python
df_can.sort_values(by='Total', ascending=False).head(5)[years].transpose().plot(kind='line', figsize=(14, 8))
plt.title('Top 5 Countries - Immigration to Canada (1980-2013)')
plt.ylabel('Number of Immigrants')
plt.xlabel('Year')
plt.show()
```

## Other Plot Types

Besides line plots, Matplotlib supports many other plot types: bar, barh, hist, box, kde, area, pie, scatter, hexbin, etc.

```python
# Example: bar plot for India
df_can.loc[['India'], years].transpose().plot(kind='bar', figsize=(12, 5))
plt.title('Bar Plot - Immigration from India (1980-2013)')
plt.ylabel('Number of Immigrants')
plt.xlabel('Year')
plt.show()
```

```python
# Example: histogram of Total column
df_can['Total'].plot(kind='hist', figsize=(8, 5))
plt.title('Histogram - Total Immigration per Country')
plt.xlabel('Total Immigrants')
plt.show()
```
