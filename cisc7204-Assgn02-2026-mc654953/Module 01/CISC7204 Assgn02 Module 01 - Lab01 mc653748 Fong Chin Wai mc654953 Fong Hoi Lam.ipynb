# Exploring and Pre-processing a Dataset using Pandas

This lab is a quick introduction to the **Pandas** library. The objective is to learn how to **pre-process and analyze a dataset** before applying data visualization techniques.

In this lab, we will explore the **Immigration to Canada from 1980 to 2013** dataset, which is obtained from the UN International migration flows - The 2015 revision.

```python
import numpy as np
import pandas as pd

print('pandas version:', pd.__version__)
```

## Read the Dataset

Use `pd.read_excel()` to load the Canada immigration dataset. The raw data is stored in the sheet `Canada by Citizenship`, with 20 rows of header notes to skip and 2 footer rows to skip.

```python
df_can = pd.read_excel('Canada.xlsx',
                       sheet_name='Canada by Citizenship',
                       skiprows=range(20),
                       skipfooter=2)

print('Data read into a pandas dataframe!')
```

## Basic Exploration

Let's get a first look at the dataframe.

```python
df_can.head()
```

```python
df_can.tail()
```

```python
df_can.info()
```

```python
# Columns, index, shape
print('Columns:', df_can.columns.tolist())
print()
print('Index:', df_can.index)
print()
print('Shape (rows, cols):', df_can.shape)
```

## Data Cleaning - Drop Unnecessary Columns

The columns `AREA`, `REG`, `DEV`, `Type`, and `Coverage` are not needed for our analysis, so we drop them.

```python
df_can.drop(columns=['AREA', 'REG', 'DEV', 'Type', 'Coverage'], inplace=True)

df_can.head(2)
```

## Rename Columns

Rename the columns to more readable names: `OdName` → `Country`, `AreaName` → `Continent`, `RegName` → `Region`.

```python
df_can.rename(columns={'OdName': 'Country',
                        'AreaName': 'Continent',
                        'RegName': 'Region'}, inplace=True)

df_can.columns
```

## Add the 'Total' Column

Create a `Total` column that sums the immigration numbers of all years (1980-2013) for each country.

```python
df_can['Total'] = df_can.sum(axis=1, numeric_only=True)

df_can.head()
```

## Check for Missing Values and Summary Statistics

```python
df_can.isnull().sum()
```

```python
df_can.describe()
```

## Indexing and Selection

Selecting columns and rows using different Pandas methods.

```python
# Column selection
df_can.Country.head()
df_can['Country'].head()
```

```python
# Row selection by label and by position
df_can.loc[3]
df_can.iloc[3]
```

```python
# Set 'Country' as the index
df_can.set_index('Country', inplace=True)

df_can.head(3)
```

### Exercise 1: Let's view the full row of **Haiti**, the data for the year **2000**, and the data for **1990-1995**.

```python
# Full row of Haiti
df_can.loc['Haiti']
```

```python
# Data for 1990-1995
df_can.loc['Haiti', 1990:1995]
```

## Convert the Column Labels to Strings

For easier plotting, convert the year column labels to strings.

```python
df_can.columns = list(map(str, df_can.columns))

df_can.columns
```

### Exercise 2: Use the `map` function to create a list `year` from 1990 to 2013 and extract the Haiti data for those years.

```python
year = list(map(str, range(1990, 2014)))

df_can.loc['Haiti', year]
```

## Conditional Filtering

Filter the dataframe based on conditions.

```python
# Countries in Asia
df_can[df_can['Continent'] == 'Asia'].head()
```

```python
# Countries in Asia AND Southern Asia region (multiple conditions with &)
df_can[(df_can['Continent'] == 'Asia') & (df_can['Region'] == 'Southern Asia')].head()
```

### Exercise 3: Extract the data for countries in **Africa** and the **Southern Africa** region, then count how many rows there are.

```python
africa_southern = df_can[(df_can['Continent'] == 'Africa') &
                        (df_can['Region'] == 'Southern Africa')]

africa_southern.head()
print('Number of rows:', len(africa_southern))
```

## Sorting Values

Sort the dataframe by the `Total` column.

```python
df_can.sort_values(by='Total', ascending=False, inplace=True)

df_can.head(5)
```

### Exercise 4: Find the **top 3 countries** that contributed the most immigrants to Canada in the year **2010**.

```python
df_can[['2010']].sort_values(by='2010', ascending=False).head(3)
```
